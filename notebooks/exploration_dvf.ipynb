```python
## Sélection du nouveau tableau 
colonnes = [
    "id_mutation", "date_mutation", "nature_mutation", "valeur_fonciere",
    "code_postal", "code_commune", "nom_commune",
    "type_local", "surface_reelle_bati", "nombre_pieces_principales",
    "surface_terrain", "longitude", "latitude",
]

dvf = df[colonnes].copy()
dvf.head()
```

```python
# Une ligne DVF = un bien, pas une vente : la valeur foncière est le prix de la mutation entière, répété sur chaque ligne qui la compose. On vérifie l'ampleur du phénomène avant tout calcul de prix.
print(dvf["id_mutation"].nunique())
dvf[dvf["id_mutation"] == dvf["id_mutation"].value_counts().index[0]]
```

```python
# Distribution du nombre de lignes par mutation
dvf["id_mutation"].value_counts().value_counts().sort_index()
```

```python
# Cas extrêmes 
dvf["id_mutation"].value_counts().head(10)
```

```python
# Identification des mutations exploitables (lignes simples)
lignes_par_mutation = dvf["id_mutation"].value_counts()
mutations_simples = lignes_par_mutation[lignes_par_mutation == 1].index
print(len(mutations_simples))
print(mutations_simples[:10])
```

```python
# Distribution des mutations 
dvf["id_mutation"].value_counts().value_counts().sort_index().head(10)
```

```python
# Compter les biens principaux par mutation
biens_principaux = dvf[dvf["type_local"].isin(["Maison", "Appartement"])]
print(len(biens_principaux))
nb_principaux = biens_principaux.groupby("id_mutation").size()
nb_principaux.value_counts().sort_index().head()
```

```python
# Ne garder les mutations qu'à un seul bien 
un_bien = nb_principaux[nb_principaux == 1].index
dvf_biens = biens_principaux[biens_principaux["id_mutation"].isin(un_bien)].copy()
dvf_biens.shape
```

```python
# Sur dvf_biens, combien de biens sont des maisons et combien sont des appartements ?
maisons = dvf_biens[dvf_biens["type_local"] == "Maison"]
print(len(maisons))

appartements = dvf_biens[dvf_biens["type_local"] == "Appartement"]
print(len(appartements))   
```

```python
# Sur dvf_biens, combien de biens ont une valeur_fonciere manquante ? 
valeur_fonciere_manquante = dvf_biens["valeur_fonciere"].isna().sum()
print(valeur_fonciere_manquante)

# Sur dvf_biens, combien de biens ont une surface_reelle_bati nulle ou manquante ?
surface_reelle_bati_manquante = dvf_biens["surface_reelle_bati"].isna().sum()
surface_reelle_bati_nulle = (dvf_biens["surface_reelle_bati"] == 0).sum()
print(surface_reelle_bati_manquante)   
print(surface_reelle_bati_nulle)
```

```python
# Valeur foncères maximum et minimum 
valeur_fonciere_max = dvf_biens["valeur_fonciere"].max()
valeur_fonciere_min = dvf_biens["valeur_fonciere"].min()
print(valeur_fonciere_max)
print(valeur_fonciere_min)
```

```python
# Création d'une nouvelle colonne "prix_m2" dans le DataFrame dvf_biens, calculée en divisant la valeur foncière par la surface réelle bâtie.
dvf_biens["prix_m2"] = dvf_biens["valeur_fonciere"] / dvf_biens["surface_reelle_bati"]
dvf_biens.head()
```

```python
# Calcul du prix médian par m2
prix_median = dvf_biens["prix_m2"].median()
print(prix_median)
```

```python
dvf_biens["prix_m2"].describe()
```

```python
# Quantiles du prix par m2 pour déterminer où se trouvent les extrêmes
dvf_biens["prix_m2"].quantile([0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99])
dvf_biens["prix_m2"].quantile([0.001, 0.01, 0.05, 0.25, 0.5, 0.75, 0.95, 0.99, 0.999])
```

```python
# Exposition des cas extrêmes
colonnes_prix_m2 = ["nom_commune", "type_local", "valeur_fonciere", "surface_reelle_bati", "prix_m2"]

dvf_biens.nsmallest(10, "prix_m2")[colonnes_prix_m2]
#dvf_biens.nlargest(10, "prix_m2")[colonnes_prix_m2]
```

```python
# Combien de biens se trouvent hors de l'intervalle [300, 15000] ?
print((dvf_biens["prix_m2"] < 300).sum())
print((dvf_biens["prix_m2"] > 15000).sum())
```

```python
# Prix au m² hors de 300–15 000 € : ventes symboliques (1 €), cessions entre proches, ou surfaces mal déclarées. Aucune transaction crédible en Ille-et-Vilaine
# n'est en dehors de cette fourchette. 115 biens écartés (0,9 %).
dvf_propre = dvf_biens[(dvf_biens["prix_m2"] >= 300) & (dvf_biens["prix_m2"] <= 15000)].copy()
dvf_propre["prix_m2"].describe()
```

```python
# Comparaison des prix au m² par type de local (Maison vs Appartement)
dvf_propre.groupby("type_local")["prix_m2"].median()
```

```python
par_commune = dvf_propre.groupby("nom_commune")["prix_m2"].agg(["count", "median"])
par_commune[par_commune["count"] > 20].nlargest(10, "median")
```

```python
par_commune[par_commune["count"] >20].nsmallest(10, "median")
```

```python
dvf_propre.head()
```

```python
dvf_communes = dvf_propre.groupby("nom_commune").agg(
    nb_ventes=("prix_m2", "count"),
    prix_median=("prix_m2", "median"),
    surface_mediane=("surface_reelle_bati", "median"),
)

dvf_communes = dvf_communes[dvf_communes["nb_ventes"] >= 5].copy()
dvf_communes.shape
```

```python
par_type = dvf_propre.pivot_table(
    index="nom_commune",
    columns="type_local",
    values="prix_m2",
    aggfunc="median",
)

par_type
```

```python
dvf_communes = dvf_communes.join(par_type)
dvf_communes = dvf_communes.rename(columns={
    "Maison": "prix_median_maison",
    "Appartement": "prix_median_appartement",
})
dvf_communes = dvf_communes.round(0)
dvf_communes
```
